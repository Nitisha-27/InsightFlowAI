Data warehouse

```python
# ============================================================
# PHASE 3
# CREATE ANALYTICS WAREHOUSE
# ============================================================

print("="*80)
print("CREATING ANALYTICS WAREHOUSE")
print("="*80)

# ------------------------------------------------------------
# Orders + Customers
# ------------------------------------------------------------

warehouse = orders_clean.merge(
    customers_clean,
    on="customer_id",
    how="left"
)

print("After Customers :", warehouse.shape)

# ------------------------------------------------------------
# Order Items
# ------------------------------------------------------------

warehouse = warehouse.merge(
    order_items_clean,
    on="order_id",
    how="left"
)

print("After Order Items :", warehouse.shape)

# ------------------------------------------------------------
# Products
# ------------------------------------------------------------

warehouse = warehouse.merge(
    products_clean,
    on="product_id",
    how="left"
)

print("After Products :", warehouse.shape)

# ------------------------------------------------------------
# Sellers
# ------------------------------------------------------------

warehouse = warehouse.merge(
    sellers_clean,
    on="seller_id",
    how="left"
)

print("After Sellers :", warehouse.shape)

# ------------------------------------------------------------
# Payments
# ------------------------------------------------------------

payment_summary = (

    payments_clean

    .groupby("order_id")

    .agg(

        total_payment=("payment_value","sum"),

        payment_type=("payment_type", lambda x: ",".join(sorted(x.astype(str).unique()))),

        payment_installments=("payment_installments","max")

    )

    .reset_index()

)

warehouse = warehouse.merge(
    payment_summary,
    on="order_id",
    how="left"
)

print("After Payments :", warehouse.shape)

# ------------------------------------------------------------
# Reviews
# ------------------------------------------------------------

review_summary = (

    reviews_clean

    .groupby("order_id")

    .agg(

        review_id=("review_id","first"),

        review_score=("review_score","mean"),

        review_comment_title=("review_comment_title","first"),

        review_comment_message=("review_comment_message","first"),

        review_creation_date=("review_creation_date","first"),

        review_answer_timestamp=("review_answer_timestamp","first")

    )

    .reset_index()

)

warehouse = warehouse.merge(
    review_summary,
    on="order_id",
    how="left"
)

print("After Reviews :", warehouse.shape)

# ------------------------------------------------------------
# DELIVERY FEATURES
# ------------------------------------------------------------

warehouse["approval_hours"] = (

    warehouse["order_approved_at"]

    -

    warehouse["order_purchase_timestamp"]

).dt.total_seconds() / 3600

warehouse["shipping_days"] = (

    warehouse["order_delivered_carrier_date"]

    -

    warehouse["order_approved_at"]

).dt.days

warehouse["delivery_days"] = (

    warehouse["order_delivered_customer_date"]

    -

    warehouse["order_delivered_carrier_date"]

).dt.days

warehouse["estimated_delivery_days"] = (

    warehouse["order_estimated_delivery_date"]

    -

    warehouse["order_purchase_timestamp"]

).dt.days

# ------------------------------------------------------------
# DATE FEATURES
# ------------------------------------------------------------

warehouse["purchase_year"] = (
    warehouse["order_purchase_timestamp"].dt.year
)

warehouse["purchase_month"] = (
    warehouse["order_purchase_timestamp"].dt.month_name()
)

warehouse["purchase_month_num"] = (
    warehouse["order_purchase_timestamp"].dt.month
)

warehouse["purchase_day"] = (
    warehouse["order_purchase_timestamp"].dt.day_name()
)

warehouse["purchase_hour"] = (
    warehouse["order_purchase_timestamp"].dt.hour
)

warehouse["purchase_quarter"] = (
    warehouse["order_purchase_timestamp"].dt.quarter
)

# ------------------------------------------------------------
# REVIEW CATEGORY
# ------------------------------------------------------------

warehouse["review_category"] = pd.cut(

    warehouse["review_score"],

    bins=[0,2,3,4,5],

    labels=[

        "Poor",

        "Average",

        "Good",

        "Excellent"

    ],

    include_lowest=True

)

# ------------------------------------------------------------
# PRODUCT FEATURES
# ------------------------------------------------------------

warehouse["product_volume_cm3"] = (

    warehouse["product_length_cm"]

    *

    warehouse["product_width_cm"]

    *

    warehouse["product_height_cm"]

)

warehouse["total_product_cost"] = (

    warehouse["price"]

    +

    warehouse["freight_value"]

)

# ------------------------------------------------------------
# LOCAL ORDER
# ------------------------------------------------------------

warehouse["local_order"] = (

    warehouse["customer_state"]

    ==

    warehouse["seller_state"]

)

print("\n")

print("="*80)
print("WAREHOUSE CREATED SUCCESSFULLY")
print("="*80)

print("Rows :", warehouse.shape[0])
print("Columns :", warehouse.shape[1])

display(warehouse.head())
```

Analytics Warehouse

```python
# ============================================================
# DATA QUALITY FIXES
# ============================================================

import numpy as np

# Invalid shipping days
warehouse.loc[
    warehouse["shipping_days"] < 0,
    "shipping_days"
] = np.nan

# Product category
warehouse["product_category"] = warehouse["product_category"].fillna("Unknown")

# Review category
warehouse["review_category"] = (
    warehouse["review_category"]
    .cat.add_categories(["No Review"])
    .fillna("No Review")
)

# Price per gram
warehouse["price_per_gram"] = (
    warehouse["price"] /
    warehouse["product_weight_g"].replace(0, np.nan)
)

# Delivery completed
warehouse["delivery_completed"] = (
    warehouse["order_delivered_customer_date"].notna()
)

# Product density
warehouse["product_density"] = (
    warehouse["product_weight_g"] /
    warehouse["product_volume_cm3"].replace(0, np.nan)
)

# ============================================================
# CUSTOMER FEATURES
# ============================================================

warehouse = warehouse.sort_values(
    ["customer_unique_id",
     "order_purchase_timestamp"]
)

warehouse["customer_order_number"] = (
    warehouse.groupby("customer_unique_id")
    .cumcount()+1
)

warehouse["repeat_customer"] = (
    warehouse["customer_order_number"] > 1
)

customer_orders = (
    warehouse.groupby("customer_unique_id")
    ["order_id"]
    .transform("nunique")
)

warehouse["customer_total_orders"] = customer_orders

warehouse["customer_lifetime_value"] = (
    warehouse.groupby("customer_unique_id")
    ["total_payment"]
    .transform("sum")
)

warehouse["customer_avg_order_value"] = (
    warehouse.groupby("customer_unique_id")
    ["total_payment"]
    .transform("mean")
)

warehouse["customer_first_purchase"] = (
    warehouse.groupby("customer_unique_id")
    ["order_purchase_timestamp"]
    .transform("min")
)

warehouse["customer_last_purchase"] = (
    warehouse.groupby("customer_unique_id")
    ["order_purchase_timestamp"]
    .transform("max")
)

warehouse["customer_tenure_days"] = (

    warehouse["customer_last_purchase"]

    -

    warehouse["customer_first_purchase"]

).dt.days

warehouse["new_customer"] = (
    warehouse["customer_order_number"] == 1
)

# ============================================================
# PRODUCT FEATURES
# ============================================================

warehouse["product_sales_count"] = (
    warehouse.groupby("product_id")
    ["order_id"]
    .transform("count")
)

warehouse["product_total_revenue"] = (
    warehouse.groupby("product_id")
    ["total_payment"]
    .transform("sum")
)

warehouse["product_avg_review"] = (
    warehouse.groupby("product_id")
    ["review_score"]
    .transform("mean")
)

warehouse["product_avg_price"] = (
    warehouse.groupby("product_id")
    ["price"]
    .transform("mean")
)

warehouse["product_popularity_rank"] = (
    warehouse["product_sales_count"]
    .rank(method="dense", ascending=False)
)

# ============================================================
# SELLER FEATURES
# ============================================================

warehouse["seller_total_orders"] = (
    warehouse.groupby("seller_id")
    ["order_id"]
    .transform("count")
)

warehouse["seller_revenue"] = (
    warehouse.groupby("seller_id")
    ["total_payment"]
    .transform("sum")
)

warehouse["seller_avg_review"] = (
    warehouse.groupby("seller_id")
    ["review_score"]
    .transform("mean")
)

warehouse["seller_rank"] = (
    warehouse["seller_revenue"]
    .rank(
        ascending=False,
        method="dense"
    )
)

# ============================================================
# DELIVERY FEATURES
# ============================================================

warehouse["delivery_delay_days"] = (

    warehouse["delivery_days"]

    -

    warehouse["estimated_delivery_days"]

)

warehouse["early_delivery"] = (
    warehouse["delivery_delay_days"] < 0
)

warehouse["on_time_delivery"] = (
    warehouse["delivery_delay_days"] <= 0
)

warehouse["late_delivery_days"] = (
    warehouse["delivery_delay_days"]
    .clip(lower=0)
)

# ============================================================
# REVENUE FEATURES
# ============================================================

warehouse["freight_percentage"] = (

    warehouse["freight_value"]

    /

    warehouse["total_product_cost"]

) * 100

warehouse["profit_proxy"] = (

    warehouse["price"]

    -

    warehouse["freight_value"]

)

warehouse["high_margin"] = (
    warehouse["profit_proxy"] >
    warehouse["profit_proxy"].median()
)

warehouse["premium_order"] = (
    warehouse["total_payment"] >
    warehouse["total_payment"].quantile(0.90)
)

# ============================================================
# FINAL FEATURE ENRICHMENT
# (Run this BEFORE saving the warehouse)
# ============================================================

print("="*80)
print("ADDING FINAL BUSINESS FEATURES")
print("="*80)

# ------------------------------------------------------------
# 1. Product Average Review
# ------------------------------------------------------------

warehouse["product_avg_review"] = (
    warehouse["product_avg_review"]
    .fillna(0)
)

# ------------------------------------------------------------
# 2. Customer Segment
# ------------------------------------------------------------

warehouse["customer_segment"] = pd.cut(

    warehouse["customer_lifetime_value"],

    bins=[0,100,500,2000,np.inf],

    labels=[

        "Bronze",

        "Silver",

        "Gold",

        "Platinum"

    ],

    include_lowest=True

)

# ------------------------------------------------------------
# 3. Product Revenue Rank
# ------------------------------------------------------------

warehouse["product_revenue_rank"] = (

    warehouse["product_total_revenue"]

    .rank(

        ascending=False,

        method="dense"

    )

)

# ------------------------------------------------------------
# 4. Basket Size
# ------------------------------------------------------------

basket_size = (

    warehouse

    .groupby("order_id")

    ["order_item_id"]

    .count()

)

warehouse["basket_size"] = (

    warehouse["order_id"]

    .map(basket_size)

)

# ------------------------------------------------------------
# 5. Basket Value
# ------------------------------------------------------------

basket_value = (

    warehouse

    .groupby("order_id")

    ["total_payment"]

    .first()

)

warehouse["basket_value"] = (

    warehouse["order_id"]

    .map(basket_value)

)

# ------------------------------------------------------------
# 6. Average Delivery Time Per Seller
# ------------------------------------------------------------

warehouse["seller_avg_delivery_days"] = (

    warehouse

    .groupby("seller_id")

    ["delivery_days"]

    .transform("mean")

)

# ------------------------------------------------------------
# 7. Seller Performance
# ------------------------------------------------------------

warehouse["seller_performance"] = pd.cut(

    warehouse["seller_avg_review"],

    bins=[0,3,4,4.5,5],

    labels=[

        "Poor",

        "Average",

        "Good",

        "Excellent"

    ],

    include_lowest=True

)

# ------------------------------------------------------------
# 8. Order Size
# ------------------------------------------------------------

warehouse["order_size"] = pd.cut(

    warehouse["basket_size"],

    bins=[0,1,3,5,100],

    labels=[

        "Single Item",

        "Small Basket",

        "Medium Basket",

        "Large Basket"

    ],

    include_lowest=True

)

# ------------------------------------------------------------
# 9. Revenue Percentile
# ------------------------------------------------------------

warehouse["revenue_percentile"] = (

    warehouse["total_payment"]

    .rank(pct=True)

)

# ------------------------------------------------------------
# 10. Top 10% Customer
# ------------------------------------------------------------

warehouse["top_customer"] = (

    warehouse["customer_lifetime_value"]

    >=

    warehouse["customer_lifetime_value"]

    .quantile(0.90)

)

# ------------------------------------------------------------
# 11. Top 10% Seller
# ------------------------------------------------------------

warehouse["top_seller"] = (

    warehouse["seller_revenue"]

    >=

    warehouse["seller_revenue"]

    .quantile(0.90)

)

# ------------------------------------------------------------
# 12. Top 10% Product
# ------------------------------------------------------------

warehouse["top_product"] = (

    warehouse["product_total_revenue"]

    >=

    warehouse["product_total_revenue"]

    .quantile(0.90)

)

# ------------------------------------------------------------
# 13. Revenue Class
# ------------------------------------------------------------

warehouse["revenue_class"] = pd.cut(

    warehouse["total_payment"],

    bins=[0,100,300,600,1000,np.inf],

    labels=[

        "Very Low",

        "Low",

        "Medium",

        "High",

        "Premium"

    ],

    include_lowest=True

)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\nFinal Warehouse Shape:", warehouse.shape)

print("\nTotal Features:", warehouse.shape[1])

print("\nNewest Features Added:")

new_features = [

    "customer_segment",

    "product_revenue_rank",

    "basket_size",

    "basket_value",

    "seller_avg_delivery_days",

    "seller_performance",

    "order_size",

    "revenue_percentile",

    "top_customer",

    "top_seller",

    "top_product",

    "revenue_class"

]

print(new_features)

print("\nFinal Feature Engineering Completed Successfully!")

# ============================================================
# TIME FEATURES
# ============================================================

warehouse["purchase_week"] = (
    warehouse["order_purchase_timestamp"]
    .dt.isocalendar()
    .week
)

warehouse["purchase_day_number"] = (
    warehouse["order_purchase_timestamp"]
    .dt.day
)

warehouse["purchase_weekend"] = (
    warehouse["order_purchase_timestamp"]
    .dt.dayofweek >= 5
)

warehouse["purchase_season"] = np.select(

    [

        warehouse["purchase_month"].isin(
            ["December","January","February"]
        ),

        warehouse["purchase_month"].isin(
            ["March","April","May"]
        ),

        warehouse["purchase_month"].isin(
            ["June","July","August"]
        ),

        warehouse["purchase_month"].isin(
            ["September","October","November"]
        )

    ],

    [

        "Summer",

        "Autumn",

        "Winter",

        "Spring"

    ],

    default="Unknown"

)

print("="*80)
print("FINAL WAREHOUSE")
print("="*80)

print("Rows :", warehouse.shape[0])

print("Columns :", warehouse.shape[1])

print("\nRemaining Missing Values")

display(

    warehouse.isnull()

    .sum()

    .sort_values(ascending=False)

    .head(20)

)

display(warehouse.head())

warehouse.to_parquet(
    "/content/drive/MyDrive/InsightFlow_AI/data/warehouse/final_analytics_warehouse.parquet",
    index=False
)

warehouse.to_csv(
    "/content/drive/MyDrive/InsightFlow_AI/data/warehouse/final_analytics_warehouse.csv",
    index=False
)

print("Final Warehouse Saved Successfully!")

```

```python
# ============================================================
# ENTERPRISE ANALYTICS WAREHOUSE
# CONFIGURATION
# ============================================================

import os
import json
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",100)

print("Libraries Loaded Successfully")

# ============================================================
# PROJECT PATHS
# ============================================================

BASE = "/content/drive/MyDrive/InsightFlow_AI"

WAREHOUSE = os.path.join(BASE,"data","warehouse")

DIMENSIONS = os.path.join(WAREHOUSE,"dimensions")

FACTS = os.path.join(WAREHOUSE,"facts")

METADATA = os.path.join(WAREHOUSE,"metadata")

REPORTS = os.path.join(BASE,"reports")

for folder in [

    DIMENSIONS,

    FACTS,

    METADATA,

    REPORTS

]:

    os.makedirs(folder,exist_ok=True)

print("Folders Created Successfully")

# ============================================================
# LOAD FINAL WAREHOUSE
# ============================================================

warehouse = pd.read_parquet(

    BASE+

    "/data/warehouse/final_analytics_warehouse.parquet"

)

print("Warehouse Loaded")

print(warehouse.shape)

print("="*80)

print("WAREHOUSE SUMMARY")

print("="*80)

print("Rows :",warehouse.shape[0])

print("Columns :",warehouse.shape[1])

print()

print("Memory Usage")

print(

warehouse.memory_usage(

deep=True

).sum()/1024**2,

"MB"

)

print()

display(

warehouse.head()

)

schema = pd.DataFrame({

"Column":warehouse.columns,

"Datatype":warehouse.dtypes.astype(str),

"Missing":

warehouse.isnull().sum(),

"Unique":

warehouse.nunique()

})

display(schema)

schema.to_csv(

METADATA+

"/warehouse_schema.csv",

index=False

)

print("Schema Saved")

dictionary = pd.DataFrame({

"Column":warehouse.columns,

"Business_Name":"",

"Description":"",

"Datatype":

warehouse.dtypes.astype(str),

"Source_Table":"",

"Business_Rule":"",

"Nullable":"",

"Primary_Key":"",

"Foreign_Key":""

})

dictionary.to_csv(

METADATA+

"/data_dictionary.csv",

index=False

)

print("Business Dictionary Template Created")

pipeline = {

"Project":"InsightFlow AI",

"Version":"1.0",

"Warehouse Rows":

int(

warehouse.shape[0]

),

"Warehouse Columns":

int(

warehouse.shape[1]

),

"Created_By":"Nitisha Mishra",

"Pipeline":"Enterprise Analytics Warehouse"

}

with open(

METADATA+

"/pipeline.json",

"w"

) as f:

    json.dump(

        pipeline,

        f,

        indent=4

    )

print("Pipeline Metadata Saved")

# ============================================================
# BUILD DIMENSION HELPER
# ============================================================

def build_dimension(
    df,
    business_key,
    surrogate_key_name,
    columns,
    sort_by=None
):
    """
    Generic dimension builder.
    Creates one row per business key and assigns
    a deterministic surrogate key.
    """

    # Keep only columns that actually exist
    available_columns = [
        col for col in columns
        if col in df.columns
    ]

    dimension = df[available_columns].copy()

    # Remove missing business keys
    dimension = dimension[
        dimension[business_key].notna()
    ].copy()

    # Remove duplicate business keys
    dimension = (
        dimension
        .sort_values(
            sort_by if sort_by else business_key
        )
        .drop_duplicates(
            subset=[business_key],
            keep="first"
        )
        .reset_index(drop=True)
    )

    # Remove old surrogate key if present
    if surrogate_key_name in dimension.columns:
        dimension.drop(
            columns=[surrogate_key_name],
            inplace=True
        )

    # Create surrogate key
    dimension.insert(
        0,
        surrogate_key_name,
        range(
            1,
            len(dimension) + 1
        )
    )

    return dimension


print("✓ build_dimension() defined successfully")


 # ============================================================
# DIM_CUSTOMERS (FIXED)
# ============================================================

customer_columns = [

    "customer_id",
    "customer_unique_id",
    "customer_city",
    "customer_state",
    "customer_zip_code_prefix",
    "customer_segment",
    "customer_total_orders",
    "customer_lifetime_value",
    "customer_avg_order_value",
    "customer_tenure_days",
    "repeat_customer"

]

dim_customers = (

    warehouse[customer_columns]

    .sort_values("customer_id")

    .drop_duplicates(
        subset=["customer_id"],
        keep="last"
    )

    .reset_index(drop=True)

)

dim_customers.insert(

    0,

    "customer_key",

    range(1, len(dim_customers)+1)

)

print("="*70)
print("DIM_CUSTOMERS")
print("="*70)
print("Rows :", len(dim_customers))
print("Columns :", len(dim_customers.columns))
print("Duplicate Customer IDs :", dim_customers["customer_id"].duplicated().sum())

display(dim_customers.head())

# ============================================================
# DIM_PRODUCTS
# ============================================================

dim_products = build_dimension(

    warehouse,

    business_key="product_id",

    surrogate_key_name="product_key",

    columns=[

        "product_id",

        "product_category",

        "product_weight_g",

        "product_length_cm",

        "product_width_cm",

        "product_height_cm",

        "product_volume_cm3",

        "product_sales_count",

        "product_total_revenue",

        "product_avg_review",

        "product_avg_price",

        "product_popularity_rank",

        "product_revenue_rank"

    ]

)

display(dim_products.head())

# ============================================================
# DIM_SELLERS
# ============================================================

dim_sellers = build_dimension(

    warehouse,

    business_key="seller_id",

    surrogate_key_name="seller_key",

    columns=[

        "seller_id",

        "seller_city",

        "seller_state",

        "seller_total_orders",

        "seller_revenue",

        "seller_avg_review",

        "seller_avg_delivery_days",

        "seller_rank",

        "seller_performance"

    ]

)

display(dim_sellers.head())

# ============================================================
# DIM_PAYMENTS
# ============================================================

# ============================================================
# DIM_PAYMENTS (FIXED)
# ============================================================

payment_summary = (

    warehouse

    .groupby("order_id")

    .agg(

        payment_type=("payment_type", "first"),

        payment_installments=("payment_installments", "max"),

        total_payment=("total_payment", "sum")

    )

    .reset_index()

)

payment_summary["payment_count"] = (

    warehouse

    .groupby("order_id")

    .size()

    .values

)

payment_summary.rename(

    columns={

        "payment_type":"payment_methods"

    },

    inplace=True

)

payment_summary.insert(

    0,

    "payment_key",

    range(1,len(payment_summary)+1)

)

dim_payments = payment_summary

print("="*70)
print("DIM_PAYMENTS")
print("="*70)

print("Rows :",len(dim_payments))
print("Columns :",len(dim_payments.columns))

display(dim_payments.head())

# ============================================================
# DIM_REVIEWS
# ============================================================

# ============================================================
# DIM_REVIEWS (FIXED)
# ============================================================

dim_reviews = (

    warehouse[[
        "order_id",
        "review_score",
        "review_category"
    ]]

    .drop_duplicates(subset=["order_id"])

    .reset_index(drop=True)

)

# Create review availability
dim_reviews["review_available"] = (
    dim_reviews["review_score"].notna()
)

# Create review label from score
dim_reviews["review_label"] = np.select(

    [

        dim_reviews["review_score"] >= 4,

        dim_reviews["review_score"] == 3,

        dim_reviews["review_score"] <= 2,

        dim_reviews["review_score"].isna()

    ],

    [

        "Positive",

        "Neutral",

        "Negative",

        "No Review"

    ],

    default="Unknown"

)

# Create surrogate key
dim_reviews.insert(

    0,

    "review_key",

    range(1, len(dim_reviews)+1)

)

print("="*70)
print("DIM_REVIEWS")
print("="*70)

print("Rows :", len(dim_reviews))
print("Columns :", len(dim_reviews.columns))
print("Duplicate Orders :", dim_reviews["order_id"].duplicated().sum())

display(dim_reviews.head())

# ============================================================
# DIM_LOCATION
# ============================================================

dim_locations = warehouse[[

    "customer_city",

    "customer_state",

    "seller_city",

    "seller_state",

    "local_order"

]].drop_duplicates()

dim_locations = dim_locations.sort_values([

    "customer_state",

    "customer_city",

    "seller_state",

    "seller_city"

]).reset_index(drop=True)

dim_locations.insert(

    0,

    "location_key",

    range(1, len(dim_locations)+1)

)

display(dim_locations.head())

# ============================================================
# DIM_ORDER_STATUS (FIXED)
# ============================================================

dim_status = (

    warehouse[[

        "order_status"

    ]]

    .drop_duplicates()

    .sort_values("order_status")

    .reset_index(drop=True)

)

dim_status.insert(

    0,

    "status_key",

    range(1, len(dim_status)+1)

)

print("="*70)
print("DIM_STATUS")
print("="*70)
print("Rows :", len(dim_status))
print("Unique Status :", dim_status["order_status"].nunique())

display(dim_status)

# ================================
# FIX PRODUCT DIMENSION
# ================================

# Remove rows with missing Product IDs
dim_products = dim_products.dropna(subset=["product_id"])

# Remove duplicate Product IDs
dim_products = (
    dim_products
    .sort_values("product_id")
    .drop_duplicates(subset=["product_id"], keep="first")
    .reset_index(drop=True)
)

# Remove old surrogate key if it already exists
if "product_key" in dim_products.columns:
    dim_products.drop(columns=["product_key"], inplace=True)

# Create new surrogate key
dim_products.insert(
    0,
    "product_key",
    range(1, len(dim_products)+1)
)

print("="*60)
print("PRODUCT DIMENSION")
print("="*60)
print("Rows :", len(dim_products))
print("Duplicate Product IDs :", dim_products["product_id"].duplicated().sum())
print("Missing Product IDs :", dim_products["product_id"].isna().sum())

# ================================
# FIX SELLER DIMENSION
# ================================

dim_sellers = dim_sellers.dropna(subset=["seller_id"])

dim_sellers = (
    dim_sellers
    .sort_values("seller_id")
    .drop_duplicates(subset=["seller_id"], keep="first")
    .reset_index(drop=True)
)

if "seller_key" in dim_sellers.columns:
    dim_sellers.drop(columns=["seller_key"], inplace=True)

dim_sellers.insert(
    0,
    "seller_key",
    range(1, len(dim_sellers)+1)
)

print("="*60)
print("SELLER DIMENSION")
print("="*60)
print("Rows :", len(dim_sellers))
print("Duplicate Seller IDs :", dim_sellers["seller_id"].duplicated().sum())
print("Missing Seller IDs :", dim_sellers["seller_id"].isna().sum())

# ============================================================
# DIM_DATE (FIXED)
# ============================================================

date_dimension = pd.DataFrame({

    "date": pd.date_range(

        warehouse["order_purchase_timestamp"].min().normalize(),

        warehouse["order_purchase_timestamp"].max().normalize(),

        freq="D"

    )

})

date_dimension["purchase_year"] = date_dimension["date"].dt.year

date_dimension["purchase_quarter"] = date_dimension["date"].dt.quarter

date_dimension["purchase_month"] = date_dimension["date"].dt.month_name()

date_dimension["purchase_week"] = (
    date_dimension["date"]
    .dt.isocalendar()
    .week
    .astype(int)
)

date_dimension["purchase_day"] = date_dimension["date"].dt.day_name()

date_dimension["purchase_day_number"] = date_dimension["date"].dt.day

date_dimension["purchase_weekend"] = (
    date_dimension["date"].dt.dayofweek >= 5
)

date_dimension["purchase_season"] = np.select(

    [

        date_dimension["date"].dt.month.isin([12,1,2]),

        date_dimension["date"].dt.month.isin([3,4,5]),

        date_dimension["date"].dt.month.isin([6,7,8]),

        date_dimension["date"].dt.month.isin([9,10,11])

    ],

    [

        "Summer",

        "Autumn",

        "Winter",

        "Spring"

    ],

    default="Unknown"

)

dim_date = date_dimension.copy()

dim_date.insert(

    0,

    "date_key",

    range(1, len(dim_date)+1)

)

print("="*70)
print("DIM_DATE")
print("="*70)
print("Rows :", len(dim_date))
print("Columns :", len(dim_date.columns))

display(dim_date.head())

# ============================================================
# SAVE DIMENSIONS
# ============================================================

dimension_tables = {

    "dim_customers": dim_customers,

    "dim_products": dim_products,

    "dim_sellers": dim_sellers,

    "dim_payments": dim_payments,

    "dim_reviews": dim_reviews,

    "dim_locations": dim_locations,

    "dim_status": dim_status,

    "dim_date": dim_date

}

for name, table in dimension_tables.items():

    table.to_parquet(
        os.path.join(DIMENSIONS, f"{name}.parquet"),
        index=False
    )

    table.to_csv(
        os.path.join(DIMENSIONS, f"{name}.csv"),
        index=False
    )

print("="*80)
print("ALL DIMENSION TABLES CREATED")
print("="*80)

for name, table in dimension_tables.items():
    print(f"{name:<20} {table.shape}")


```

```python
# ============================================================
# PHASE 8
# BUSINESS INTELLIGENCE & AUTOMATED INSIGHT ENGINE
# ============================================================

import os
import sqlite3
import pandas as pd
import numpy as np

# ============================================================
# CONFIGURATION
# ============================================================

BASE = "/content/drive/MyDrive/InsightFlow_AI"

DB_PATH = os.path.join(
    BASE,
    "warehouse.db"
)

PHASE8_PATH = os.path.join(
    BASE,
    "reports",
    "phase8_business_intelligence"
)

os.makedirs(
    PHASE8_PATH,
    exist_ok=True
)

print("=" * 90)
print("INSIGHTFLOW AI")
print("PHASE 8 — BUSINESS INTELLIGENCE & INSIGHT ENGINE")
print("=" * 90)

# ============================================================
# CONNECT TO SQLITE
# ============================================================

conn = sqlite3.connect(DB_PATH)

print("\nSQLite Connected Successfully!")

# ============================================================
# VERIFY DATABASE
# ============================================================

required_tables = [
    "fact_sales",
    "dim_customers",
    "dim_products",
    "dim_sellers",
    "dim_reviews",
    "dim_payments",
    "dim_locations",
    "dim_status",
    "dim_date"
]

existing_tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    """,
    conn
)["name"].tolist()

missing_tables = [
    table
    for table in required_tables
    if table not in existing_tables
]

if missing_tables:

    conn.close()

    raise Exception(
        "Missing required tables: "
        + ", ".join(missing_tables)
    )

print("All required warehouse tables found.")

# ============================================================
# HELPER
# ============================================================

def save_output(df, filename):

    csv_path = os.path.join(
        PHASE8_PATH,
        filename + ".csv"
    )

    parquet_path = os.path.join(
        PHASE8_PATH,
        filename + ".parquet"
    )

    df.to_csv(
        csv_path,
        index=False
    )

    df.to_parquet(
        parquet_path,
        index=False
    )

    return csv_path


def run_phase8_query(
    title,
    query,
    filename
):

    print("\n")
    print("=" * 90)
    print(title)
    print("=" * 90)

    result = pd.read_sql_query(
        query,
        conn
    )

    display(result)

    save_output(
        result,
        filename
    )

    print(
        f"Saved → {filename}.csv"
    )

    return result


# ============================================================
# 8.1 EXECUTIVE KPI SCORECARD
# ============================================================

executive_kpis = run_phase8_query(

    "8.1 EXECUTIVE KPI SCORECARD",

    """

    SELECT

        COUNT(DISTINCT order_id)
            AS TotalOrders,

        COUNT(DISTINCT customer_key)
            AS TotalCustomers,

        COUNT(DISTINCT product_key)
            AS TotalProducts,

        COUNT(DISTINCT seller_key)
            AS TotalSellers,

        ROUND(
            SUM(total_payment),
            2
        ) AS TotalRevenue,

        ROUND(
            SUM(total_payment)
            /
            NULLIF(
                COUNT(DISTINCT order_id),
                0
            ),
            2
        ) AS AverageOrderValue,

        ROUND(
            AVG(basket_size),
            2
        ) AS AverageBasketSize,

        ROUND(
            AVG(review_score),
            2
        ) AS AverageReviewScore,

        ROUND(
            AVG(delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            AVG(shipping_days),
            2
        ) AS AverageShippingDays,

        ROUND(
            SUM(
                CASE
                    WHEN delivery_delay_days > 0
                    THEN 1
                    ELSE 0
                END
            ) * 100.0
            /
            NULLIF(COUNT(*),0),
            2
        ) AS LateDeliveryRate,

        ROUND(
            SUM(profit_proxy),
            2
        ) AS ProfitProxy,

        ROUND(
            SUM(profit_proxy) * 100.0
            /
            NULLIF(
                SUM(total_payment),
                0
            ),
            2
        ) AS ProfitMarginPercent

    FROM fact_sales

    """,

    "executive_kpis"

)


# ============================================================
# 8.2 MONTHLY BUSINESS PERFORMANCE
# ============================================================

monthly_business = run_phase8_query(

    "8.2 MONTHLY BUSINESS PERFORMANCE",

    """

    WITH monthly AS (

        SELECT

            d.purchase_year AS Year,

            d.purchase_month AS Month,

            SUM(f.total_payment)
                AS Revenue,

            COUNT(
                DISTINCT f.order_id
            ) AS Orders,

            COUNT(
                DISTINCT f.customer_key
            ) AS Customers,

            SUM(f.profit_proxy)
                AS ProfitProxy

        FROM fact_sales f

        INNER JOIN dim_date d

            ON f.date_key = d.date_key

        GROUP BY

            d.purchase_year,

            d.purchase_month

    )

    SELECT

        Year,

        Month,

        ROUND(
            Revenue,
            2
        ) AS Revenue,

        Orders,

        Customers,

        ROUND(
            Revenue
            /
            NULLIF(Orders,0),
            2
        ) AS AverageOrderValue,

        ROUND(
            ProfitProxy,
            2
        ) AS ProfitProxy,

        ROUND(
            ProfitProxy * 100.0
            /
            NULLIF(Revenue,0),
            2
        ) AS ProfitMarginPercent,

        ROUND(

            (
                Revenue
                -
                LAG(Revenue)
                OVER (
                    ORDER BY Year, Month
                )

            ) * 100.0

            /

            NULLIF(

                LAG(Revenue)
                OVER (
                    ORDER BY Year, Month
                ),

                0

            ),

            2

        ) AS RevenueGrowthPercent

    FROM monthly

    ORDER BY

        Year,

        Month

    """,

    "monthly_business_performance"

)


# ============================================================
# 8.3 PRODUCT INTELLIGENCE
# ============================================================

product_intelligence = run_phase8_query(

    "8.3 PRODUCT INTELLIGENCE",

    """

    SELECT

        p.product_id,

        p.product_category,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        SUM(
            f.order_item_id
        ) AS Quantity,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            SUM(f.profit_proxy),
            2
        ) AS ProfitProxy,

        ROUND(
            AVG(f.review_score),
            2
        ) AS AverageReview,

        ROUND(
            AVG(f.price),
            2
        ) AS AveragePrice,

        ROUND(
            SUM(f.profit_proxy) * 100.0
            /
            NULLIF(
                SUM(f.total_payment),
                0
            ),
            2
        ) AS ProfitMarginPercent

    FROM fact_sales f

    INNER JOIN dim_products p

        ON f.product_key = p.product_key

    WHERE

        p.product_id != 'UNKNOWN'

    GROUP BY

        p.product_id,

        p.product_category

    ORDER BY

        Revenue DESC

    """,

    "product_intelligence"

)


# ============================================================
# 8.4 PRODUCT CATEGORY INTELLIGENCE
# ============================================================

category_intelligence = run_phase8_query(

    "8.4 PRODUCT CATEGORY INTELLIGENCE",

    """

    SELECT

        p.product_category,

        COUNT(
            DISTINCT p.product_id
        ) AS Products,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            SUM(f.profit_proxy),
            2
        ) AS ProfitProxy,

        ROUND(
            AVG(f.review_score),
            2
        ) AS AverageReview,

        ROUND(
            AVG(f.delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            SUM(f.profit_proxy) * 100.0
            /
            NULLIF(
                SUM(f.total_payment),
                0
            ),
            2
        ) AS ProfitMarginPercent

    FROM fact_sales f

    INNER JOIN dim_products p

        ON f.product_key = p.product_key

    WHERE

        p.product_category IS NOT NULL

        AND p.product_category != 'Unknown'

    GROUP BY

        p.product_category

    ORDER BY

        Revenue DESC

    """,

    "category_intelligence"

)


# ============================================================
# 8.5 CUSTOMER INTELLIGENCE
# ============================================================

customer_intelligence = run_phase8_query(

    "8.5 CUSTOMER INTELLIGENCE",

    """

    SELECT

        customer_key,

        COUNT(
            DISTINCT order_id
        ) AS Orders,

        ROUND(
            SUM(total_payment),
            2
        ) AS LifetimeValue,

        ROUND(
            SUM(total_payment)
            /
            NULLIF(
                COUNT(DISTINCT order_id),
                0
            ),
            2
        ) AS AverageOrderValue,

        ROUND(
            AVG(review_score),
            2
        ) AS AverageReview,

        ROUND(
            AVG(delivery_days),
            2
        ) AS AverageDeliveryDays,

        CASE

            WHEN COUNT(
                DISTINCT order_id
            ) >= 5

            AND SUM(total_payment) >= 1000

                THEN 'VIP'

            WHEN COUNT(
                DISTINCT order_id
            ) >= 3

                THEN 'High Value'

            WHEN COUNT(
                DISTINCT order_id
            ) = 2

                THEN 'Regular'

            ELSE 'One Time'

        END AS CustomerSegment

    FROM fact_sales

    WHERE

        customer_key != 0

    GROUP BY

        customer_key

    ORDER BY

        LifetimeValue DESC

    """,

    "customer_intelligence"

)


# ============================================================
# 8.6 CUSTOMER SEGMENT SUMMARY
# ============================================================

customer_segments = run_phase8_query(

    "8.6 CUSTOMER SEGMENT SUMMARY",

    """

    WITH customers AS (

        SELECT

            customer_key,

            COUNT(
                DISTINCT order_id
            ) AS Orders,

            SUM(total_payment)
                AS LifetimeValue

        FROM fact_sales

        WHERE customer_key != 0

        GROUP BY customer_key

    )

    SELECT

        CASE

            WHEN Orders >= 5
            AND LifetimeValue >= 1000
                THEN 'VIP'

            WHEN Orders >= 3
                THEN 'High Value'

            WHEN Orders = 2
                THEN 'Regular'

            ELSE 'One Time'

        END AS CustomerSegment,

        COUNT(*) AS Customers,

        ROUND(
            SUM(LifetimeValue),
            2
        ) AS Revenue,

        ROUND(
            AVG(LifetimeValue),
            2
        ) AS AverageCustomerValue

    FROM customers

    GROUP BY

        CASE

            WHEN Orders >= 5
            AND LifetimeValue >= 1000
                THEN 'VIP'

            WHEN Orders >= 3
                THEN 'High Value'

            WHEN Orders = 2
                THEN 'Regular'

            ELSE 'One Time'

        END

    ORDER BY

        Revenue DESC

    """,

    "customer_segment_summary"

)


# ============================================================
# 8.7 SELLER INTELLIGENCE
# ============================================================

seller_intelligence = run_phase8_query(

    "8.7 SELLER INTELLIGENCE",

    """

    SELECT

        s.seller_id,

        s.seller_state,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            SUM(f.profit_proxy),
            2
        ) AS ProfitProxy,

        ROUND(
            AVG(f.review_score),
            2
        ) AS AverageReview,

        ROUND(
            AVG(f.delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            SUM(f.profit_proxy) * 100.0
            /
            NULLIF(
                SUM(f.total_payment),
                0
            ),
            2
        ) AS ProfitMarginPercent

    FROM fact_sales f

    INNER JOIN dim_sellers s

        ON f.seller_key = s.seller_key

    WHERE

        s.seller_id != 'UNKNOWN'

    GROUP BY

        s.seller_id,

        s.seller_state

    ORDER BY

        Revenue DESC

    """,

    "seller_intelligence"

)


# ============================================================
# 8.8 SELLER PERFORMANCE MATRIX
# ============================================================

seller_performance = run_phase8_query(

    "8.8 SELLER PERFORMANCE MATRIX",

    """

    WITH seller_metrics AS (

        SELECT

            s.seller_id,

            COUNT(
                DISTINCT f.order_id
            ) AS Orders,

            SUM(
                f.total_payment
            ) AS Revenue,

            AVG(
                f.review_score
            ) AS ReviewScore,

            AVG(
                f.delivery_days
            ) AS DeliveryDays

        FROM fact_sales f

        INNER JOIN dim_sellers s

            ON f.seller_key =
               s.seller_key

        WHERE

            s.seller_id != 'UNKNOWN'

        GROUP BY

            s.seller_id

    ),

    thresholds AS (

        SELECT

            AVG(Revenue)
                AS AvgRevenue,

            AVG(ReviewScore)
                AS AvgReview

        FROM seller_metrics

    )

    SELECT

        s.seller_id,

        s.Orders,

        ROUND(
            s.Revenue,
            2
        ) AS Revenue,

        ROUND(
            s.ReviewScore,
            2
        ) AS ReviewScore,

        ROUND(
            s.DeliveryDays,
            2
        ) AS DeliveryDays,

        CASE

            WHEN s.Revenue >= t.AvgRevenue
            AND s.ReviewScore >= t.AvgReview

                THEN 'High Performer'

            WHEN s.Revenue >= t.AvgRevenue
            AND s.ReviewScore < t.AvgReview

                THEN 'Revenue Leader - Review Risk'

            WHEN s.Revenue < t.AvgRevenue
            AND s.ReviewScore >= t.AvgReview

                THEN 'Quality Opportunity'

            ELSE 'Needs Improvement'

        END AS SellerSegment

    FROM seller_metrics s

    CROSS JOIN thresholds t

    ORDER BY

        Revenue DESC

    """,

    "seller_performance_matrix"

)


# ============================================================
# 8.9 DELIVERY INTELLIGENCE
# ============================================================

delivery_intelligence = run_phase8_query(

    "8.9 DELIVERY INTELLIGENCE",

    """

    SELECT

        CASE

            WHEN delivery_delay_days > 0
                THEN 'Late'

            ELSE 'On Time'

        END AS DeliveryStatus,

        COUNT(*) AS Orders,

        ROUND(
            COUNT(*) * 100.0
            /
            NULLIF(
                (
                    SELECT COUNT(*)
                    FROM fact_sales
                ),
                0
            ),
            2
        ) AS OrderSharePercent,

        ROUND(
            AVG(delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            AVG(shipping_days),
            2
        ) AS AverageShippingDays,

        ROUND(
            AVG(delivery_delay_days),
            2
        ) AS AverageDelayDays,

        ROUND(
            AVG(review_score),
            2
        ) AS AverageReview

    FROM fact_sales

    GROUP BY

        CASE

            WHEN delivery_delay_days > 0
                THEN 'Late'

            ELSE 'On Time'

        END

    """,

    "delivery_intelligence"

)


# ============================================================
# 8.10 GEOGRAPHIC INTELLIGENCE
# ============================================================

geographic_intelligence = run_phase8_query(

    "8.10 GEOGRAPHIC INTELLIGENCE",

    """

    SELECT

        l.customer_state AS State,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        COUNT(
            DISTINCT f.customer_key
        ) AS Customers,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            SUM(f.profit_proxy),
            2
        ) AS ProfitProxy,

        ROUND(
            AVG(f.review_score),
            2
        ) AS AverageReview,

        ROUND(
            AVG(f.delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            SUM(
                CASE
                    WHEN f.delivery_delay_days > 0
                    THEN 1
                    ELSE 0
                END
            ) * 100.0
            /
            NULLIF(COUNT(*),0),
            2
        ) AS LateDeliveryRate

    FROM fact_sales f

    INNER JOIN dim_locations l

        ON f.location_key =
           l.location_key

    GROUP BY

        l.customer_state

    ORDER BY

        Revenue DESC

    """,

    "geographic_intelligence"

)


# ============================================================
# 8.11 LOCAL VS NON-LOCAL
# ============================================================

local_delivery_analysis = run_phase8_query(

    "8.11 LOCAL VS NON-LOCAL DELIVERY",

    """

    SELECT

        CASE

            WHEN l.local_order = 1
                THEN 'Local'

            ELSE 'Non-Local'

        END AS OrderType,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            AVG(f.delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            AVG(f.shipping_days),
            2
        ) AS AverageShippingDays,

        ROUND(
            AVG(f.review_score),
            2
        ) AS AverageReview,

        ROUND(
            SUM(
                CASE
                    WHEN f.delivery_delay_days > 0
                    THEN 1
                    ELSE 0
                END
            ) * 100.0
            /
            NULLIF(COUNT(*),0),
            2
        ) AS LateDeliveryRate

    FROM fact_sales f

    INNER JOIN dim_locations l

        ON f.location_key =
           l.location_key

    GROUP BY

        l.local_order

    ORDER BY

        AverageDeliveryDays

    """,

    "local_vs_nonlocal_delivery"

)


# ============================================================
# 8.12 PAYMENT INTELLIGENCE
# ============================================================

payment_intelligence = run_phase8_query(

    "8.12 PAYMENT INTELLIGENCE",

    """

    SELECT

        p.payment_methods AS PaymentMethod,

        COUNT(
            DISTINCT f.order_id
        ) AS Orders,

        ROUND(
            SUM(f.total_payment),
            2
        ) AS Revenue,

        ROUND(
            SUM(f.total_payment)
            /
            NULLIF(
                COUNT(DISTINCT f.order_id),
                0
            ),
            2
        ) AS AverageOrderValue,

        ROUND(
            AVG(
                p.payment_installments
            ),
            2
        ) AS AverageInstallments,

        ROUND(
            AVG(
                p.payment_count
            ),
            2
        ) AS AveragePaymentCount

    FROM fact_sales f

    INNER JOIN dim_payments p

        ON f.payment_key =
           p.payment_key

    GROUP BY

        p.payment_methods

    ORDER BY

        Revenue DESC

    """,

    "payment_intelligence"

)


# ============================================================
# 8.13 REVIEW / CUSTOMER EXPERIENCE
# ============================================================

review_intelligence = run_phase8_query(

    "8.13 REVIEW & CUSTOMER EXPERIENCE",

    """

    SELECT

        CASE

            WHEN review_score >= 4
                THEN 'Positive'

            WHEN review_score = 3
                THEN 'Neutral'

            WHEN review_score <= 2
                THEN 'Negative'

            ELSE 'No Review'

        END AS ReviewSentiment,

        COUNT(
            DISTINCT order_id
        ) AS Orders,

        ROUND(
            SUM(total_payment),
            2
        ) AS Revenue,

        ROUND(
            AVG(delivery_days),
            2
        ) AS AverageDeliveryDays,

        ROUND(
            AVG(delivery_delay_days),
            2
        ) AS AverageDelayDays

    FROM fact_sales

    GROUP BY

        CASE

            WHEN review_score >= 4
                THEN 'Positive'

            WHEN review_score = 3
                THEN 'Neutral'

            WHEN review_score <= 2
                THEN 'Negative'

            ELSE 'No Review'

        END

    ORDER BY

        Revenue DESC

    """,

    "review_customer_experience"

)


# ============================================================
# 8.14 AUTOMATED BUSINESS INSIGHTS
# ============================================================

print("\n")
print("=" * 90)
print("8.14 AUTOMATED BUSINESS INSIGHT GENERATION")
print("=" * 90)

insights = []


# ------------------------------------------------------------
# Executive KPI insights
# ------------------------------------------------------------

kpi = executive_kpis.iloc[0]

total_revenue = float(
    kpi["TotalRevenue"]
)

total_orders = int(
    kpi["TotalOrders"]
)

aov = float(
    kpi["AverageOrderValue"]
)

late_rate = float(
    kpi["LateDeliveryRate"]
)

profit_margin = float(
    kpi["ProfitMarginPercent"]
)

insights.append({

    "InsightType":
        "Executive",

    "Priority":
        "High",

    "Metric":
        "Revenue",

    "Insight":
        f"Total revenue is {total_revenue:,.2f} "
        f"across {total_orders:,} orders.",

    "Recommendation":
        "Use revenue and order trends as the primary "
        "executive performance indicators."

})


insights.append({

    "InsightType":
        "Executive",

    "Priority":
        "High",

    "Metric":
        "Average Order Value",

    "Insight":
        f"Average order value is {aov:,.2f}.",

    "Recommendation":
        "Increase basket size through cross-selling, "
        "bundling and personalized recommendations."

})


insights.append({

    "InsightType":
        "Operations",

    "Priority":
        "High" if late_rate >= 5 else "Medium",

    "Metric":
        "Late Delivery Rate",

    "Insight":
        f"Late delivery rate is {late_rate:.2f}%.",

    "Recommendation":
        "Investigate logistics, seller processing time "
        "and destination-level delivery bottlenecks."

})


insights.append({

    "InsightType":
        "Profitability",

    "Priority":
        "High",

    "Metric":
        "Profit Margin",

    "Insight":
        f"Current profit proxy margin is "
        f"{profit_margin:.2f}%.",

    "Recommendation":
        "Monitor pricing, freight cost and product mix "
        "to protect contribution margins."

})


# ------------------------------------------------------------
# Monthly trend insight
# ------------------------------------------------------------

if len(monthly_business) >= 2:

    latest = monthly_business.iloc[-1]

    previous = monthly_business.iloc[-2]

    growth = latest["RevenueGrowthPercent"]

    if pd.notna(growth):

        direction = (
            "increased"
            if growth >= 0
            else "decreased"
        )

        insights.append({

            "InsightType":
                "Revenue Trend",

            "Priority":
                "High",

            "Metric":
                "Monthly Revenue Growth",

            "Insight":
                f"Revenue {direction} by "
                f"{abs(growth):.2f}% "
                f"from the previous month.",

            "Recommendation":
                "Investigate the drivers of the latest "
                "monthly revenue movement."

        })


# ------------------------------------------------------------
# Product insight
# ------------------------------------------------------------

if len(category_intelligence) > 0:

    top_category = (
        category_intelligence
        .iloc[0]
    )

    insights.append({

        "InsightType":
            "Product",

        "Priority":
            "High",

        "Metric":
            "Top Category",

        "Insight":
            f"{top_category['product_category']} "
            f"is the highest-revenue product category "
            f"with revenue of "
            f"{top_category['Revenue']:,.2f}.",

        "Recommendation":
            "Protect inventory availability and analyze "
            "cross-selling opportunities around this category."

    })


# ------------------------------------------------------------
# Seller insight
# ------------------------------------------------------------

if len(seller_performance) > 0:

    high_performers = seller_performance[
        seller_performance["SellerSegment"]
        == "High Performer"
    ]

    insights.append({

        "InsightType":
            "Seller",

        "Priority":
            "Medium",

        "Metric":
            "High Performing Sellers",

        "Insight":
            f"{len(high_performers):,} sellers "
            f"are classified as High Performers.",

        "Recommendation":
            "Study high-performing seller practices "
            "and replicate them across weaker sellers."

    })


# ------------------------------------------------------------
# Customer insight
# ------------------------------------------------------------

if len(customer_segments) > 0:

    vip = customer_segments[
        customer_segments["CustomerSegment"]
        == "VIP"
    ]

    if len(vip) > 0:

        vip_customers = int(
            vip.iloc[0]["Customers"]
        )

        vip_revenue = float(
            vip.iloc[0]["Revenue"]
        )

        insights.append({

            "InsightType":
                "Customer",

            "Priority":
                "High",

            "Metric":
                "VIP Customers",

            "Insight":
                f"{vip_customers:,} customers "
                f"are classified as VIP and generate "
                f"{vip_revenue:,.2f} in revenue.",

            "Recommendation":
                "Use retention campaigns and personalized "
                "offers to protect high-value customers."

        })


# ------------------------------------------------------------
# Delivery insight
# ------------------------------------------------------------

if len(local_delivery_analysis) >= 2:

    local_row = local_delivery_analysis[
        local_delivery_analysis["OrderType"]
        == "Local"
    ]

    nonlocal_row = local_delivery_analysis[
        local_delivery_analysis["OrderType"]
        == "Non-Local"
    ]

    if (
        len(local_row) > 0
        and len(nonlocal_row) > 0
    ):

        local_days = float(
            local_row.iloc[0][
                "AverageDeliveryDays"
            ]
        )

        nonlocal_days = float(
            nonlocal_row.iloc[0][
                "AverageDeliveryDays"
            ]
        )

        if local_days > 0:

            ratio = (
                nonlocal_days
                /
                local_days
            )

            insights.append({

                "InsightType":
                    "Logistics",

                "Priority":
                    "High",

                "Metric":
                    "Local vs Non-Local",

                "Insight":
                    f"Non-local orders take approximately "
                    f"{ratio:.2f}× the delivery time of "
                    f"local orders.",

                "Recommendation":
                    "Optimize inter-state logistics, "
                    "carrier allocation and fulfillment routing."

            })


# ------------------------------------------------------------
# Payment insight
# ------------------------------------------------------------

if len(payment_intelligence) > 0:

    top_payment = (
        payment_intelligence
        .iloc[0]
    )

    insights.append({

        "InsightType":
            "Payments",

        "Priority":
            "Medium",

        "Metric":
            "Top Payment Method",

        "Insight":
            f"{top_payment['PaymentMethod']} "
            f"generates the highest revenue at "
            f"{top_payment['Revenue']:,.2f}.",

        "Recommendation":
            "Prioritize reliability and conversion "
            "optimization for the dominant payment method."

    })


# ============================================================
# SAVE BUSINESS INSIGHTS
# ============================================================

business_insights = pd.DataFrame(
    insights
)

display(
    business_insights
)

save_output(
    business_insights,
    "business_insights"
)


# ============================================================
# 8.15 CREATE PHASE 8 MASTER SUMMARY
# ============================================================

phase8_summary = pd.DataFrame({

    "Component": [

        "Executive KPI",

        "Monthly Business Performance",

        "Product Intelligence",

        "Category Intelligence",

        "Customer Intelligence",

        "Customer Segmentation",

        "Seller Intelligence",

        "Seller Performance Matrix",

        "Delivery Intelligence",

        "Geographic Intelligence",

        "Local vs Non-Local",

        "Payment Intelligence",

        "Review Intelligence",

        "Automated Business Insights"

    ],

    "Status": [

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed"

    ]

})

display(
    phase8_summary
)

save_output(
    phase8_summary,
    "phase8_summary"
)


# ============================================================
# 8.16 FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 90)
print("PHASE 8 COMPLETED SUCCESSFULLY")
print("=" * 90)

print("\nOutput Directory:")
print(PHASE8_PATH)

print("\nFiles Generated:")

for file in sorted(
    os.listdir(PHASE8_PATH)
):

    print("✓", file)

print("\n")
print("=" * 90)
print("BUSINESS INTELLIGENCE LAYER READY")
print("=" * 90)

# ============================================================
# CLOSE CONNECTION
# ============================================================

conn.close()

print("\nSQLite connection closed.")
print("InsightFlow AI Phase 8 is ready for Phase 9.")
```

```python
# ============================================================
# INSIGHTFLOW AI
# PHASE 10B
# PRODUCTION-GRADE CUSTOMER RETENTION ENGINE
# ============================================================

import pandas as pd
import numpy as np
import os
import warnings

warnings.filterwarnings("ignore")

print("="*90)
print("INSIGHTFLOW AI — PHASE 10B")
print("ROLLING CUSTOMER RETENTION PREDICTION")
print("="*90)

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/InsightFlow_AI"

PHASE10B_PATH = os.path.join(
    BASE,
    "reports",
    "phase10b_retention"
)

os.makedirs(PHASE10B_PATH, exist_ok=True)

# ============================================================
# PREPARE ORDER-LEVEL DATA
# ============================================================

orders_ml = warehouse.copy()

orders_ml["order_purchase_timestamp"] = pd.to_datetime(
    orders_ml["order_purchase_timestamp"],
    errors="coerce"
)

orders_ml = orders_ml[
    orders_ml["order_purchase_timestamp"].notna()
].copy()

# ============================================================
# IMPORTANT:
# CUSTOMER IDENTITY = customer_unique_id
# This represents the actual customer across orders.
# ============================================================

orders_ml = orders_ml[
    orders_ml["customer_unique_id"].notna()
].copy()

orders_ml = orders_ml.sort_values(
    "order_purchase_timestamp"
).reset_index(drop=True)

print("="*90)
print("ORDER DATA READY")
print("="*90)

print("Rows              :", len(orders_ml))
print(
    "Unique Orders     :",
    orders_ml["order_id"].nunique()
)
print(
    "Unique Customers  :",
    orders_ml["customer_unique_id"].nunique()
)

# ============================================================
# TEMPORAL RANGE
# ============================================================

min_date = orders_ml["order_purchase_timestamp"].min()
max_date = orders_ml["order_purchase_timestamp"].max()

print()
print("Earliest transaction :", min_date)
print("Latest transaction   :", max_date)

# ============================================================
# PARAMETERS
# ============================================================

OBSERVATION_DAYS = 180
PREDICTION_DAYS = 90
STEP_DAYS = 30

print()
print("Observation Window :", OBSERVATION_DAYS, "days")
print("Prediction Window   :", PREDICTION_DAYS, "days")
print("Snapshot Frequency  :", STEP_DAYS, "days")


# ============================================================
# 10B.2
# ROLLING CUSTOMER SNAPSHOTS
# ============================================================

snapshot_start = (
    min_date
    + pd.Timedelta(days=OBSERVATION_DAYS)
)

snapshot_end = (
    max_date
    - pd.Timedelta(days=PREDICTION_DAYS)
)

snapshot_dates = pd.date_range(
    start=snapshot_start.normalize(),
    end=snapshot_end.normalize(),
    freq=f"{STEP_DAYS}D"
)

print("="*90)
print("CREATING ROLLING SNAPSHOTS")
print("="*90)

print("Number of snapshots:", len(snapshot_dates))
print("First snapshot      :", snapshot_dates.min())
print("Last snapshot       :", snapshot_dates.max())

# ============================================================
# FEATURE COLUMNS
# ============================================================

feature_columns = [
    "TotalOrders",
    "TotalRevenue",
    "AverageOrderValue",
    "MaximumOrderValue",
    "AverageBasketSize",
    "AverageBasketValue",
    "TotalItems",
    "AverageFreight",
    "AverageDeliveryDays",
    "AverageReview",
    "TotalProfitProxy",
    "RecencyDays",
    "CustomerLifetimeDays",
    "PurchaseFrequency",
    "OrderVelocity",
    "FreightRatio",
    "CategoryDiversity",
    "SellerDiversity",
    "WeekendPurchaseRatio"
]

snapshot_records = []

# ============================================================
# CREATE SNAPSHOTS
# ============================================================

for i, snapshot_date in enumerate(snapshot_dates):

    observation_start = (
        snapshot_date
        - pd.Timedelta(days=OBSERVATION_DAYS)
    )

    prediction_end = (
        snapshot_date
        + pd.Timedelta(days=PREDICTION_DAYS)
    )

    # --------------------------------------------------------
    # Historical data only
    # --------------------------------------------------------

    historical = orders_ml[
        (orders_ml["order_purchase_timestamp"] >= observation_start)
        &
        (orders_ml["order_purchase_timestamp"] < snapshot_date)
    ].copy()

    # --------------------------------------------------------
    # Future data ONLY for target
    # --------------------------------------------------------

    future = orders_ml[
        (orders_ml["order_purchase_timestamp"] >= snapshot_date)
        &
        (orders_ml["order_purchase_timestamp"] < prediction_end)
    ].copy()

    if historical.empty:
        continue

    # --------------------------------------------------------
    # Historical customer aggregation
    # --------------------------------------------------------

    customer_features = (

        historical

        .groupby("customer_unique_id")

        .agg(

            TotalOrders=("order_id", "nunique"),

            TotalRevenue=("total_payment", "sum"),

            AverageOrderValue=("total_payment", "mean"),

            MaximumOrderValue=("total_payment", "max"),

            AverageBasketSize=("basket_size", "mean"),

            AverageBasketValue=("basket_value", "mean"),

            TotalItems=("order_item_id", "count"),

            AverageFreight=("freight_value", "mean"),

            AverageDeliveryDays=("delivery_days", "mean"),

            AverageReview=("review_score", "mean"),

            TotalProfitProxy=("profit_proxy", "sum"),

            FirstPurchase=(
                "order_purchase_timestamp",
                "min"
            ),

            LastPurchase=(
                "order_purchase_timestamp",
                "max"
            )

        )

        .reset_index()

    )

    # --------------------------------------------------------
    # Recency
    # --------------------------------------------------------

    customer_features["RecencyDays"] = (

        snapshot_date
        - customer_features["LastPurchase"]

    ).dt.total_seconds() / 86400

    # --------------------------------------------------------
    # Customer lifetime
    # --------------------------------------------------------

    customer_features["CustomerLifetimeDays"] = (

        customer_features["LastPurchase"]
        - customer_features["FirstPurchase"]

    ).dt.total_seconds() / 86400

    customer_features["CustomerLifetimeDays"] = (
        customer_features["CustomerLifetimeDays"]
        .clip(lower=0)
    )

    # --------------------------------------------------------
    # Purchase frequency
    # --------------------------------------------------------

    customer_features["PurchaseFrequency"] = (

        customer_features["TotalOrders"]
        /
        (
            customer_features["CustomerLifetimeDays"]
            + 1
        )

    )

    # --------------------------------------------------------
    # Order velocity
    # --------------------------------------------------------

    customer_features["OrderVelocity"] = (

        customer_features["TotalOrders"]
        /
        OBSERVATION_DAYS

    )

    # --------------------------------------------------------
    # Freight ratio
    # --------------------------------------------------------

    customer_features["FreightRatio"] = (

        customer_features["AverageFreight"]
        /
        customer_features["AverageOrderValue"].replace(
            0,
            np.nan
        )

    ).fillna(0)

    # --------------------------------------------------------
    # Category diversity
    # --------------------------------------------------------

    category_diversity = (

        historical

        .groupby("customer_unique_id")[
            "product_category"
        ]

        .nunique()

    )

    customer_features["CategoryDiversity"] = (

        customer_features["customer_unique_id"]
        .map(category_diversity)
        .fillna(0)

    )

    # --------------------------------------------------------
    # Seller diversity
    # --------------------------------------------------------

    seller_diversity = (

        historical

        .groupby("customer_unique_id")[
            "seller_id"
        ]

        .nunique()

    )

    customer_features["SellerDiversity"] = (

        customer_features["customer_unique_id"]
        .map(seller_diversity)
        .fillna(0)

    )

    # --------------------------------------------------------
    # Weekend purchase ratio
    # --------------------------------------------------------

    historical["purchase_day"] = (
        historical["order_purchase_timestamp"]
        .dt.dayofweek
    )

    weekend_ratio = (

        historical.assign(
            weekend=
            historical["purchase_day"].isin([5, 6])
        )

        .groupby("customer_unique_id")["weekend"]

        .mean()

    )

    customer_features["WeekendPurchaseRatio"] = (

        customer_features["customer_unique_id"]
        .map(weekend_ratio)
        .fillna(0)

    )

    # --------------------------------------------------------
    # TARGET
    #
    # Did customer purchase during next 90 days?
    # --------------------------------------------------------

    future_customers = set(
        future["customer_unique_id"].unique()
    )

    customer_features["RepeatPurchase90D"] = (

        customer_features["customer_unique_id"]
        .isin(future_customers)
        .astype(int)

    )

    # --------------------------------------------------------
    # Snapshot identifier
    # --------------------------------------------------------

    customer_features["SnapshotDate"] = snapshot_date

    snapshot_records.append(
        customer_features
    )

    if (i + 1) % 5 == 0:

        print(
            f"Processed snapshot "
            f"{i+1}/{len(snapshot_dates)}"
        )

# ============================================================
# COMBINE
# ============================================================

retention_dataset = pd.concat(
    snapshot_records,
    ignore_index=True
)

print()
print("="*90)
print("ROLLING RETENTION DATASET CREATED")
print("="*90)

print(
    "Rows              :",
    len(retention_dataset)
)

print(
    "Customers         :",
    retention_dataset["customer_unique_id"].nunique()
)

print(
    "Snapshots         :",
    retention_dataset["SnapshotDate"].nunique()
)

print(
    "Positive targets  :",
    retention_dataset["RepeatPurchase90D"].sum()
)

print(
    "Positive rate     :",
    round(
        retention_dataset["RepeatPurchase90D"].mean(),
        4
    )
)

display(
    retention_dataset.head()
)

# ============================================================
# 10B.3
# DATASET VALIDATION
# ============================================================

print("="*90)
print("RETENTION DATASET VALIDATION")
print("="*90)

print(
    "Total rows:",
    len(retention_dataset)
)

print(
    "Duplicate customer-snapshot pairs:",
    retention_dataset.duplicated(
        subset=[
            "customer_unique_id",
            "SnapshotDate"
        ]
    ).sum()
)

print(
    "Missing target:",
    retention_dataset[
        "RepeatPurchase90D"
    ].isna().sum()
)

print()
print("TARGET DISTRIBUTION")

display(
    retention_dataset[
        "RepeatPurchase90D"
    ].value_counts()
    .rename_axis("RepeatPurchase90D")
    .reset_index(name="Customers")
)

print()
print("TARGET RATE BY SNAPSHOT")

snapshot_target = (

    retention_dataset

    .groupby("SnapshotDate")

    ["RepeatPurchase90D"]

    .agg(
        Customers="count",
        Repeaters="sum",
        RepeatRate="mean"
    )

    .reset_index()

)

display(snapshot_target)

# ============================================================
# 10B.4
# SAVE ROLLING DATASET
# ============================================================

retention_dataset.to_parquet(
    os.path.join(
        PHASE10B_PATH,
        "rolling_retention_dataset.parquet"
    ),
    index=False
)

retention_dataset.to_csv(
    os.path.join(
        PHASE10B_PATH,
        "rolling_retention_dataset.csv"
    ),
    index=False
)

print("="*90)
print("ROLLING RETENTION DATASET SAVED")
print("="*90)

print(
    os.path.join(
        PHASE10B_PATH,
        "rolling_retention_dataset.parquet"
    )
)

# ============================================================
# PHASE 10B.5
# TEMPORAL TRAIN / VALIDATION / TEST SPLIT
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("="*90)
print("INSIGHTFLOW AI — PHASE 10B.5")
print("TEMPORAL MODEL SPLIT")
print("="*90)

# ============================================================
# SORT BY SNAPSHOT DATE
# ============================================================

model_data = retention_dataset.copy()

model_data["SnapshotDate"] = pd.to_datetime(
    model_data["SnapshotDate"]
)

model_data = model_data.sort_values(
    "SnapshotDate"
).reset_index(drop=True)

# ============================================================
# GET SNAPSHOT DATES
# ============================================================

dates = sorted(
    model_data["SnapshotDate"].unique()
)

n_dates = len(dates)

train_end_index = int(n_dates * 0.60)
validation_end_index = int(n_dates * 0.80)

train_dates = dates[:train_end_index]

validation_dates = dates[
    train_end_index:validation_end_index
]

test_dates = dates[
    validation_end_index:
]

# ============================================================
# SPLIT
# ============================================================

train = model_data[
    model_data["SnapshotDate"].isin(train_dates)
].copy()

validation = model_data[
    model_data["SnapshotDate"].isin(validation_dates)
].copy()

test = model_data[
    model_data["SnapshotDate"].isin(test_dates)
].copy()

print()
print("="*90)
print("TEMPORAL SPLIT")
print("="*90)

print(
    "Training snapshots   :",
    len(train_dates)
)

print(
    "Validation snapshots :",
    len(validation_dates)
)

print(
    "Test snapshots       :",
    len(test_dates)
)

print()
print(
    "Training period   :",
    train["SnapshotDate"].min(),
    "→",
    train["SnapshotDate"].max()
)

print(
    "Validation period :",
    validation["SnapshotDate"].min(),
    "→",
    validation["SnapshotDate"].max()
)

print(
    "Test period       :",
    test["SnapshotDate"].min(),
    "→",
    test["SnapshotDate"].max()
)

print()
print("Rows")
print("Training   :", len(train))
print("Validation :", len(validation))
print("Test       :", len(test))

print()
print("Positive examples")

print(
    "Training   :",
    train["RepeatPurchase90D"].sum()
)

print(
    "Validation :",
    validation["RepeatPurchase90D"].sum()
)

print(
    "Test       :",
    test["RepeatPurchase90D"].sum()
)

# ============================================================
# PHASE 10B.6
# FEATURE PREPARATION
# ============================================================

TARGET = "RepeatPurchase90D"

ID_COLUMNS = [
    "customer_unique_id",
    "SnapshotDate",
    "FirstPurchase",
    "LastPurchase"
]

FEATURES = [
    "TotalOrders",
    "TotalRevenue",
    "AverageOrderValue",
    "MaximumOrderValue",
    "AverageBasketSize",
    "AverageBasketValue",
    "TotalItems",
    "AverageFreight",
    "AverageDeliveryDays",
    "AverageReview",
    "TotalProfitProxy",
    "RecencyDays",
    "CustomerLifetimeDays",
    "PurchaseFrequency",
    "OrderVelocity",
    "FreightRatio",
    "CategoryDiversity",
    "SellerDiversity",
    "WeekendPurchaseRatio"
]

# ============================================================
# CREATE X / Y
# ============================================================

X_train = train[FEATURES].copy()
y_train = train[TARGET].copy()

X_validation = validation[FEATURES].copy()
y_validation = validation[TARGET].copy()

X_test = test[FEATURES].copy()
y_test = test[TARGET].copy()

# ============================================================
# HANDLE INFINITE VALUES
# ============================================================

for df in [
    X_train,
    X_validation,
    X_test
]:

    df.replace(
        [np.inf, -np.inf],
        np.nan,
        inplace=True
    )

# ============================================================
# MEDIAN IMPUTATION
# ============================================================

train_medians = X_train.median()

X_train = X_train.fillna(train_medians)

X_validation = X_validation.fillna(train_medians)

X_test = X_test.fillna(train_medians)

print("="*90)
print("FEATURE DATA READY")
print("="*90)

print("Number of features:", len(FEATURES))

print()
print("Training shape   :", X_train.shape)
print("Validation shape :", X_validation.shape)
print("Test shape       :", X_test.shape)

print()
print("Features:")

for feature in FEATURES:
    print("✓", feature)

    # ============================================================
# PHASE 10B.7
# OPTUNA-TUNED RANDOM FOREST
# ============================================================

!pip install -q optuna

import optuna
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

print("="*90)
print("INSIGHTFLOW AI — PHASE 10B.7")
print("OPTUNA-TUNED RANDOM FOREST")
print("="*90)

# ============================================================
# OBJECTIVE FUNCTION
# ============================================================

def objective(trial):

    params = {

        "n_estimators": trial.suggest_int(
            "n_estimators",
            200,
            700,
            step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            4,
            18
        ),

        "min_samples_split": trial.suggest_int(
            "min_samples_split",
            2,
            30
        ),

        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf",
            1,
            20
        ),

        "max_features": trial.suggest_categorical(
            "max_features",
            ["sqrt", "log2", 0.5, 0.8]
        ),

        "max_samples": trial.suggest_float(
            "max_samples",
            0.5,
            1.0
        ),

        "class_weight": trial.suggest_categorical(
            "class_weight",
            ["balanced", "balanced_subsample"]
        ),

        "random_state": 42,

        "n_jobs": -1
    }

    model = RandomForestClassifier(
        **params
    )

    model.fit(
        X_train,
        y_train
    )

    validation_probability = model.predict_proba(
        X_validation
    )[:, 1]

    # --------------------------------------------------------
    # PRIMARY OPTIMIZATION METRIC
    # --------------------------------------------------------

    pr_auc = average_precision_score(
        y_validation,
        validation_probability
    )

    return pr_auc


# ============================================================
# CREATE OPTUNA STUDY
# ============================================================

study = optuna.create_study(
    direction="maximize",
    study_name="InsightFlow_Retention_RF"
)

study.optimize(
    objective,
    n_trials=30,
    show_progress_bar=True
)

# ============================================================
# BEST PARAMETERS
# ============================================================

print()
print("="*90)
print("OPTUNA RESULTS")
print("="*90)

print(
    "Best Validation PR-AUC:",
    round(study.best_value, 6)
)

print()
print("Best Parameters:")

for key, value in study.best_params.items():

    print(
        f"{key:<25}: {value}"
    )

    # ============================================================
# 10B.7.1
# TRAIN BEST RANDOM FOREST
# ============================================================

best_rf = RandomForestClassifier(

    **study.best_params,

    random_state=42,

    n_jobs=-1
)

best_rf.fit(
    X_train,
    y_train
)

# ============================================================
# VALIDATION PREDICTIONS
# ============================================================

validation_probability = best_rf.predict_proba(
    X_validation
)[:, 1]

validation_prediction = (
    validation_probability >= 0.5
).astype(int)

validation_roc = roc_auc_score(
    y_validation,
    validation_probability
)

validation_pr = average_precision_score(
    y_validation,
    validation_probability
)

validation_precision = precision_score(
    y_validation,
    validation_prediction,
    zero_division=0
)

validation_recall = recall_score(
    y_validation,
    validation_prediction,
    zero_division=0
)

validation_f1 = f1_score(
    y_validation,
    validation_prediction,
    zero_division=0
)

print("="*90)
print("BEST RANDOM FOREST — VALIDATION RESULTS")
print("="*90)

print(
    "ROC-AUC   :",
    round(validation_roc, 6)
)

print(
    "PR-AUC    :",
    round(validation_pr, 6)
)

print(
    "Precision :",
    round(validation_precision, 6)
)

print(
    "Recall    :",
    round(validation_recall, 6)
)

print(
    "F1 Score  :",
    round(validation_f1, 6)
)





```

```python
# ============================================================
# PHASE 10B.8
# OPTUNA-TUNED XGBOOST
# ============================================================

!pip install -q xgboost

import xgboost as xgb
import optuna

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

print("="*90)
print("INSIGHTFLOW AI — PHASE 10B.8")
print("OPTUNA-TUNED XGBOOST")
print("="*90)

# ============================================================
# CALCULATE CLASS IMBALANCE
# ============================================================

negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = (
    negative_count / positive_count
)

print()
print("Negative samples :", negative_count)
print("Positive samples :", positive_count)

print(
    "Scale Pos Weight :",
    round(scale_pos_weight, 2)
)

# ============================================================
# OPTUNA OBJECTIVE
# ============================================================

def xgb_objective(trial):

    params = {

        "n_estimators": trial.suggest_int(
            "n_estimators",
            200,
            800,
            step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            2,
            8
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.01,
            0.15,
            log=True
        ),

        "min_child_weight": trial.suggest_int(
            "min_child_weight",
            1,
            20
        ),

        "subsample": trial.suggest_float(
            "subsample",
            0.6,
            1.0
        ),

        "colsample_bytree": trial.suggest_float(
            "colsample_bytree",
            0.6,
            1.0
        ),

        "gamma": trial.suggest_float(
            "gamma",
            0,
            5
        ),

        "reg_alpha": trial.suggest_float(
            "reg_alpha",
            0,
            5
        ),

        "reg_lambda": trial.suggest_float(
            "reg_lambda",
            0.1,
            10,
            log=True
        ),

        "scale_pos_weight": scale_pos_weight,

        "objective": "binary:logistic",

        "eval_metric": "aucpr",

        "tree_method": "hist",

        "random_state": 42,

        "n_jobs": -1
    }

    model = xgb.XGBClassifier(
        **params
    )

    model.fit(
        X_train,
        y_train,

        eval_set=[
            (X_validation, y_validation)
        ],

        verbose=False
    )

    probability = model.predict_proba(
        X_validation
    )[:, 1]

    pr_auc = average_precision_score(
        y_validation,
        probability
    )

    return pr_auc


# ============================================================
# CREATE STUDY
# ============================================================

xgb_study = optuna.create_study(
    direction="maximize",
    study_name="InsightFlow_Retention_XGB"
)

xgb_study.optimize(
    xgb_objective,
    n_trials=30,
    show_progress_bar=True
)

# ============================================================
# RESULTS
# ============================================================

print()
print("="*90)
print("XGBOOST OPTUNA RESULTS")
print("="*90)

print(
    "Best Validation PR-AUC:",
    round(
        xgb_study.best_value,
        6
    )
)

print()
print("Best Parameters:")

for key, value in xgb_study.best_params.items():

    print(
        f"{key:<25}: {value}"
    )

    # ============================================================
# 10B.8.1
# BEST XGBOOST VALIDATION MODEL
# ============================================================

best_xgb = xgb.XGBClassifier(

    **xgb_study.best_params,

    scale_pos_weight=scale_pos_weight,

    objective="binary:logistic",

    eval_metric="aucpr",

    tree_method="hist",

    random_state=42,

    n_jobs=-1
)

best_xgb.fit(
    X_train,
    y_train,

    eval_set=[
        (X_validation, y_validation)
    ],

    verbose=False
)

# ============================================================
# VALIDATION PREDICTIONS
# ============================================================

xgb_validation_probability = (
    best_xgb
    .predict_proba(X_validation)[:, 1]
)

xgb_validation_prediction = (
    xgb_validation_probability >= 0.5
).astype(int)

xgb_roc = roc_auc_score(
    y_validation,
    xgb_validation_probability
)

xgb_pr = average_precision_score(
    y_validation,
    xgb_validation_probability
)

xgb_precision = precision_score(
    y_validation,
    xgb_validation_prediction,
    zero_division=0
)

xgb_recall = recall_score(
    y_validation,
    xgb_validation_prediction,
    zero_division=0
)

xgb_f1 = f1_score(
    y_validation,
    xgb_validation_prediction,
    zero_division=0
)

print("="*90)
print("BEST XGBOOST — VALIDATION RESULTS")
print("="*90)

print("ROC-AUC   :", round(xgb_roc, 6))
print("PR-AUC    :", round(xgb_pr, 6))
print("Precision :", round(xgb_precision, 6))
print("Recall    :", round(xgb_recall, 6))
print("F1 Score  :", round(xgb_f1, 6))

# ============================================================
# 10B.8.2
# MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({

    "Model": [
        "Random Forest",
        "XGBoost"
    ],

    "ROC_AUC": [
        validation_roc,
        xgb_roc
    ],

    "PR_AUC": [
        validation_pr,
        xgb_pr
    ],

    "Precision": [
        validation_precision,
        xgb_precision
    ],

    "Recall": [
        validation_recall,
        xgb_recall
    ],

    "F1": [
        validation_f1,
        xgb_f1
    ]

})

comparison = comparison.sort_values(
    "PR_AUC",
    ascending=False
).reset_index(drop=True)

print("="*90)
print("MODEL COMPARISON")
print("="*90)

display(
    comparison.round(6)
)

# ============================================================
# SELECT WINNER
# ============================================================

winner = comparison.iloc[0]["Model"]

print()
print("="*90)
print("MODEL SELECTED")
print("="*90)

print(
    "Winner based on validation PR-AUC:",
    winner
)

```
